```python
df_full = df_h.copy()
#i taked a copy😎
print(df_full)
```

```python

# تحويل المتغيرات النصية إلى Dummy Variables (One-Hot Encoding)
df_encoded = pd.get_dummies(df_full)
#               print(df_encoded)باديه

# ملء القيم المفقودة بـ 0 (لأن المسافة لا تعمل مع NaN)
df_encoded = df_encoded.fillna(0) 
#    2. عمل Scaling للبيانات
scaler = MinMaxScaler()
df_scaled_array = scaler.fit_transform(df_encoded)
print(df_scaled_array,'\n',"*"*90)
df_scaled = pd.DataFrame(df_scaled_array, index=df_encoded.index, columns=df_encoded.columns)
print(df_scaled)
```

```python

# 3. حساب المسافة بين المنزل 0 وجميع المنازل الأخرى
target_house = df_scaled.loc[[0]] # المنزل رقم 0
distances = ps(df_scaled, target_house) # حساب المسافة للكل
print(f"target_house \n{target_house} \n {"*"*20} distances{"*"*20}:\n{distances} ")
```

```python
# إضافة عمود المسافة والسعر الحقيقي للبيانات للمقارنة
results = df_h.copy()
results['distance_to_0'] = distances
results = results[['SalePrice', 'distance_to_0', 'Gr Liv Area', 'Neighborhood']]
print(results)
```

```python
 # 4. التصفية: نشتي منازل مشابهة (مسافة قريبة) ولكن أرخص الظروf 😂
house_0_price = results.loc[0, 'SalePrice']
print(f"Price of House 0: {house_0_price}")
```

```python
# ترتيب حسب المسافة (الأقرب فالأبعد) وفلترة الأرخص
similar_cheaper_houses = results[results['SalePrice'] < house_0_price].sort_values('distance_to_0')
similar_cheaper_houses
#4 it's enough as i want.(•_•)
```
